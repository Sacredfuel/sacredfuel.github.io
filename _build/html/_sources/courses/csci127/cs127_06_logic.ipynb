## Part I. Truth Tables

Last week, we introduced decisions (if-statements). These relied on the "test" which was an expression that evaluated to true or false. 

We can build up more complicated logical expressions using built-in operators, `and`, `or`, and `not`. A standard way to summaries these operators is with truth tables. A truth table lists all the possible inputs and gives the output for each combination. 

For example, here is the truth table for the and operator: 

$$
\begin{array}{c|c|c}
in1 & in2 & in1 \text{ and } in2 \\
\hline
T & T & T \\
T & F & F \\
F & T & F \\
F & F & F \\
\end{array}
$$

It lists all the possible combinations of values for the inputs, `in1` and `in2` and what and would give as output. For example, if `in1` is True and `in2` is False, the value of `in1 and in2` can be read off the last column of the second line of the truth table as False.

## Part III. Unix

UNIX is an innovative or groundbreaking operating system which was developed in the 1970s by Ken Thompson, Dennis Ritchie, and many others at AT&T Laboratories. It is like a backbone for many modern operating systems like Ubuntu, Solaris, Kali Linux, Arch Linux, and also POSIX.

The basic design philosophy of UNIX is to provide simple, powerful tools that can be combined to perform complex tasks. It features a command-line interface (CLI) that allows users to interact with the system through a series of commands, rather than through a graphical user interface (GUI). 

| Command | What it does                                  | Acronym meaning |
|---------|-----------------------------------------------|-----------------|
| ls      | prints all the files in the current directory |                |
| cd      | opens a folder / navigates into               |                 |
| pwd     | prints the current directory                  |                 |
| mv      | moves a file / renames it                     |                 |
| cp      | makes a copy of a file                        |                 |
| wc -l   | gets how many files are currently printed     | N/A             |
|mkdir |  makes a folder                  |    make directory        |
